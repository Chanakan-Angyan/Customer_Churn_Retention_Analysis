## 10. Key Findings

### 1. Overall Churn
- Overall churn rate was **26.54%** across 7,043 customers.

### 2. Contract Type
- Month-to-month customers had the highest churn rate at **45.84%**.
- Two-year contract customers had the lowest churn rate at **2.55%**.

### 3. Customer Tenure
- Customers with **0–12 months** of tenure had the highest churn rate at **47.44%**.
- Customers with **61+ months** of tenure had the lowest churn rate at **6.61%**.

### 4. Internet Service
- Customers with Internet Service had a churn rate of **31.83%**, compared with **7.40%** among customers without Internet Service.
- Fiber Optic customers showed the highest churn rate among Internet Types at **40.72%**.

### 5. Monthly Charges
- Churn rates were higher among customers with monthly charges of **$60–100**.
- The **$80–100** group had the highest churn rate at **37.02%**.

### 6. Churn Categories
- **Competitor** was the largest reported churn category, accounting for approximately **45%** of churned customers.

### 7. Churn Reasons
- Competitor-related reasons were among the most frequently reported churn reasons.
- Support and service-related reasons were also observed among churned customers.

> These findings describe patterns and associations observed in the dataset and do not imply causal relationships.
